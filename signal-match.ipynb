```python
import numpy as np
import matplotlib.pyplot as plt

# 定义时间范围
t = np.linspace(0, 12.56, 1000)

# 定义信号x(t)
x = np.where(t <= 2 * np.pi, np.sin(t), np.sin(2 * t))

# 绘制信号
plt.figure(figsize=(8, 4))
plt.plot(t, x, label='x(t)')
plt.xlabel('t (s)')  # 时间（秒）
plt.ylabel('x(t)')   # 信号值
plt.title('Signal x(t)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# t2 is the first quarter of t
t2 = t[0:len(t)//4]

# define signal template r(t) = sin(2t)
r = np.sin(2 * t2)
r2 = np.cos(2 * t2)

# plot x(t) and r(t) in one figure
plt.figure(figsize=(8, 4))
plt.plot(t, x, color='blue', label='x(t)')
plt.plot(t2, r, color='red', label='r(t) = sin(2t)')
plt.plot(t2, r2, color='green', label='r(t) = cos(2t)')
plt.xlabel('t (s)')  # time (seconds)
plt.ylabel('Signal value')
plt.title('x(t) and Template r(t)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# 使用np.correlate计算x和r的相关性
corr = np.correlate(x, r, mode='same')/len(r)
corr2 = np.correlate(x, r2, mode='same')/len(r2)


# 绘制相关性结果
plt.figure(figsize=(8, 4))
plt.plot(t, corr, label='Correlation')
plt.plot(t, corr2, label='Correlation2')
plt.xlabel('t (s)')  # 时间（秒）
plt.ylabel('Correlation value')  # 相关性值
plt.title('Signal Matching Result')
plt.legend()
plt.grid(True)
plt.show()
```

```python
corr3 = np.sqrt(np.power(corr, 2) + np.power(corr2, 2))
plt.plot(t, corr3, label='Correlation3')
plt.xlabel('t (s)')  # 时间（秒）
plt.ylabel('Correlation value')  # 相关性值
plt.title('Signal Matching Result')
plt.legend()
plt.grid(True)
```
