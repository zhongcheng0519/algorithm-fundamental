```python
import numpy as np
import matplotlib.pyplot as plt  # import here for plotting

t = np.linspace(0, 2 * np.pi, 500)

# Generate x1(t) = sin(t) with noise
x1 = np.sin(t) + np.random.normal(0, 0.01, t.shape)

# Generate x2(t) = sin(t) with noise
x2 = -2*np.sin(t) + np.random.normal(0, 0.01, t.shape)

plt.figure(figsize=(10, 4))
plt.plot(t, x1, label='x1(t)')  # plot x1
plt.plot(t, x2, label='x2(t)')  # plot x2
plt.xlabel('t')  # x-axis label
plt.ylabel('Signal')  # y-axis label
plt.title('x1(t) and x2(t) with noise')  # plot title
plt.legend()  # show legend
plt.show()  # display plot
```

```python
# method 1: elementwise multiply and sum
match_score = np.sum(x1 * x2)  # 逐元素相乘后求和，得到匹配分数
print("Match score (method 1):", match_score)  # 输出匹配分数
```

```python
# method 2: match_score divided by norm(x1) and norm(x2)
norm_x1 = np.linalg.norm(x1)  # calculate norm of x1
norm_x2 = np.linalg.norm(x2)  # calculate norm of x2
match_score2 = match_score / (norm_x1 * norm_x2)  # normalized match score
print("Match score (method 2):", match_score2)  # output normalized match score
```

```python
# method 1 (using np.correlate): elementwise multiply and sum via correlation
match_score_corr = np.correlate(x1, x2)[0]  # use np.correlate to compute match score
print("Match score (method 1, np.correlate):", match_score_corr)  # output match score
```

```python
# method 2 (using np.corrcoef): use Pearson correlation coefficient
pearson_score = np.corrcoef(x1, x2)[0, 1]  # use np.corrcoef to calculate Pearson correlation
print("Match score (method 3, Pearson):", pearson_score)  # output Pearson correlation score

```
